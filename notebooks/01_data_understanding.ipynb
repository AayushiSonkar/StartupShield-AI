```python
target = df["labels"]
```

```python
columns_to_drop = [
    "Unnamed: 0",
    "Unnamed: 6",
    "id",
    "object_id",
    "name",
    "city",
    "zip_code",
    "state_code.1",
    "status",
    "labels",
    "closed_at",
    "founded_at",
    "first_funding_at",
    "last_funding_at"
]
```

```python
features = df.drop(columns=columns_to_drop)
```

```python
features.shape
```

```python
features.isnull().sum()
```

```python
features["age_first_milestone_year"] = features["age_first_milestone_year"].fillna(
    features["age_first_milestone_year"].median()
)
```

```python
features.isnull().sum()
```

```python
features["age_last_milestone_year"] = features["age_last_milestone_year"].fillna(
    features["age_last_milestone_year"].median()
)
```

```python
features.isnull().sum()
```

```python
features.select_dtypes(include="object").columns.tolist()
```

```python
features_encoded = pd.get_dummies(features, drop_first=True, dtype=int)
```

```python
features_encoded.shape
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    features_encoded,
    target,
    test_size=0.20,
    random_state=42,
    stratify=target
)
```

```python
print(X_train.shape)
print(X_test.shape)
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
model = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    class_weight="balanced"
)
```

```python
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", accuracy)
```

```python
print(classification_report(y_test, y_pred))
```

```python
cm = confusion_matrix(y_test, y_pred)

print(cm)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.figure(figsize=(6, 4))
```

```python
sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["Closed", "Acquired"],
    yticklabels=["Closed", "Acquired"]
)
```

```python
plt.xlabel("Predicted Label")
plt.ylabel("Actual Label")
plt.title("Random Forest Confusion Matrix")
plt.show()
```

```python
importance_df = pd.DataFrame({
    "feature": features_encoded.columns,
    "importance": model.feature_importances_
})
```

```python
importance_df = importance_df.sort_values(
    by="importance",
    ascending=False
)
```

```python
importance_df.head(10)
```

```python
top_features = importance_df.head(10)

plt.figure(figsize=(10, 6))
sns.barplot(
    data=top_features,
    x="importance",
    y="feature"
)
plt.title("Top 10 Important Startup Features")
plt.show()
```
